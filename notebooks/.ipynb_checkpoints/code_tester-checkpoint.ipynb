## Data Cleaning

```python
from pathlib import Path
import os
import re
```

```python
RAW_DIR = Path('data_raw')
CLEAN_DIR = Path('data_clean')
CLEAN_DIR.mkdir(exist_ok=True)
```

```python
path = "../data_raw"
dir_list = os.listdir(path)
```

```python
combined_df = pd.DataFrame()
pattern = re.compile(r'^(\d{4}-\d{4})-(.+?)-Stats\.csv$')

df_list = []
for f in dir_list:
    path = "../data_raw/" + f
    season = pattern.match(f).group(1)
    league = pattern.match(f).group(2)
    cur_df = pd.read_csv(path)
    cur_df['season'] = season
    cur_df['league'] = league
    df_list.append(cur_df)
    
combined_df = pd.concat(df_list, ignore_index=True)
```

```python
# combined_df.isnull().sum()
```

```python
combined_df.columns
```

```python
combined_df.dtypes
```

```python
combined_df['minutes'].value_counts()
```

```python
def convert_minutes(x):
    if pd.isnull(x):
        return pd.NA
    x = x.replace(',', "")
    return int(x)
```

```python
combined_df['minutes'] = combined_df['minutes'].apply(convert_minutes)
combined_df['minutes'].value_counts()
```

```python
combined_df.league.value_counts()
```

```python
len(set(combined_df.columns)) == len(combined_df.columns)
```

```python
duplicate_cols_to_drop = [col for col in combined_df.columns if col.endswith('_y')]
duplicate_cols_to_drop
```

```python
combined_df = combined_df.drop(columns=duplicate_cols_to_drop)
combined_df
```

```python
combined_df.columns = combined_df.columns.str.replace(r'_x$', '', regex=True)
combined_df.columns
```

```python
core = combined_df[["league","season","team","games_played","fouls","cards_yellow","cards_red"]].copy()
core["fouls_per_match_team"] = core["fouls"] / core["games_played"]

league_season = (
    core.groupby(["league","season"], as_index=False)
        .agg(total_fouls=("fouls","sum"),
             total_team_games=("games_played","sum"))
        .assign(matches=lambda d: d["total_team_games"]/2.0,
                fouls_per_match=lambda d: d["total_fouls"]/d["matches"])
        .drop(columns="total_team_games")
)
```

## SQL Setup

```python
import pandas as pd
import os
import sys
from pathlib import Path
from sqlalchemy import create_engine, text
```

```python
from dotenv import load_dotenv
```

```python
SCHEMA_SQL = "../sql/schema.sql"
INPUT_DATA = "../data_clean/fbref_stats_cleaned.csv"

load_dotenv()

SQL_HOST = os.getenv("MYSQL_HOST")
SQL_PORT = int(os.getenv("MYSQL_PORT"))
SQL_USER = os.getenv("MYSQL_USER")
SQL_PASSWORD = os.getenv("MYSQL_PASSWORD")
SQL_DB = os.getenv("MYSQL_DB")
```

```python
df = pd.read_csv(INPUT_DATA)
df.head()
```

```python
engine = create_engine(
        f"mysql+pymysql://{SQL_USER}:{SQL_PASSWORD}@{SQL_HOST}:{SQL_PORT}",
        future=True,
        pool_pre_ping=True
    )
engine
```

```python
with engine.begin() as conn:
    conn.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS {SQL_DB}"
        )
    )
```

```python
with engine.begin() as conn:
    conn.execute(text("DROP DATABASE football"))
```

```python
with engine.begin() as conn:
    sql = Path(SCHEMA_SQL).read_text(encoding="utf-8")

    commands = [s.strip() for s in sql.split(';') if s.strip()]

    for command in commands:
        conn.execute(text(command))
```

```python
with engine.begin() as conn:
    test = conn.execute(text("SHOW TABLES"))
for row in test.fetchall():
    print(row)
```

```python
with engine.begin() as conn:
        # League
        league_names = sorted(df["league"].dropna().unique())
        for name in league_names:
            t = text("""INSERT INTO League (name)
                     VALUES (:name) AS new
                     ON DUPLICATE KEY UPDATE name = new.name""")
            conn.execute(t, {"name": name})

        # Season
        seasons = sorted(df["season"].dropna().unique())
        for season in seasons:
            t = text("""
                     INSERT INTO Season (label, start_year, end_year)
                     VALUES (:season, :start, :end) AS new
                     ON DUPLICATE KEY UPDATE start_year = new.start_year, end_year = new.end_year
                     """)
            conn.execute(t, {"season": season,
                             "start": int(season.split("-")[0]),
                             "end": int(season.split("-")[1])})
            
        # Team
        teams = sorted(df["team"].dropna().unique())
        for team in teams:
            t = text("""INSERT INTO Team (name)
                     VALUES (:name) AS new
                     ON DUPLICATE KEY UPDATE name = new.name
                     """)
            conn.execute(t, {"name": team})
```

```python
def run_command(e, s):
    with e.begin() as conn:
        res = conn.execute(text(s))
    return res.fetchall()
```

```python
with engine.begin() as conn:
        league_season = df[["league", "season"]].drop_duplicates()
        for idx, row in league_season.iterrows():
            t = text("""
                     INSERT IGNORE INTO League_Season (league_id, season_id)
                     SELECT L.league_id, S.season_id
                     FROM League L JOIN Season S ON S.label=:season
                     WHERE L.name=:league
                     """)
            conn.execute(t, {"league": row["league"], "season": row["season"]})
```

```python
metric_cols = [
    'games_played', 'players_used', 'avg_age', 'possession', 
    'goals', 'assists', 'pens_made', 'pens_attempted', 'cards_yellow',
    'cards_red', 'fouls', 'fouled', 'offsides', 'crosses', 'interceptions',
    'tackles_won', 'pens_won', 'pens_conceded', 'own_goals',
    'xg', 'npxg', 'xg_assist', 'progressive_passes',
    'progressive_carries', 'ball_recoveries', 'aerials_won', 'aerials_lost'
]

# if column is missing, create column
for col in metric_cols:
    if col not in df.columns:
        df[col] = pd.NA

with engine.begin() as conn:
    for idx, row in df.iterrows():
        t = text(f"""
                 INSERT INTO Metric (ls_id, team_id, {", ".join(metric_cols)})
                 SELECT LS.ls_id, T.team_id, {", ".join([f":{col}" for col in metric_cols])}
                 FROM League_Season LS
                 JOIN League L ON LS.league_id = L.league_id
                 JOIN Season S ON LS.season_id = S.season_id
                 JOIN Team T ON T.name = :team
                 WHERE L.name = :league AND S.label = :season
                 ON DUPLICATE KEY UPDATE {", ".join([f"{col} = VALUES({col})" for col in metric_cols])}
                 """)
        
        params = {
            "league": row["league"], "season": row["season"], "team": row["team"],
            **{col: (None if pd.isnull(row[col]) else row[col]) for col in metric_cols}
        }
        conn.execute(t, params)
```
